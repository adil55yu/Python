Task 1 — Solve a New Cryptarithmetic Puzzle    

Solve the following puzzle manually (using pen and paper) and then
verify your answer with the Python code:    
        
B A S E    
+ B A L L    
─────────    
G A M E S

Questions to answer:             
a) List all unique variables and their initial domains.                         
b) Identify all constraints.                   
c) Show at least three backtracking steps in a small search tree.                 
d) State the final digit assignments and verify the sum.      

```python
import itertools

letters = ["B", "A", "S", "E", "L", "G", "M"]

solutions = []

for values in itertools.permutations(range(10), 7):

    B, A, S, E, L, G, M = values

    if B == 0 or G == 0:
        continue

    BASE = 1000 * B + 100 * A + 10 * S + E
    BALL = 1000 * B + 100 * A + 10 * L + L
    GAMES = 10000 * G + 1000 * A + 100 * M + 10 * E + S

    if BASE - BALL == GAMES:
        solutions.append({
            "B": B,
            "A": A,
            "S": S,
            "E": E,
            "L": L,
            "G": G,
            "M": M
        })

print("Number of solutions:", len(solutions))

if solutions:
    for solution in solutions:
        print(solution)
else:
    print("No solution exists.")
```

Task 2 — Modify the Code

Extend the provided Python program to solve the following puzzle:

T W O     
+ T W O    
────────    
F O U R            

Requirements:                                                  
a) Identify the unique letters: T, W, O, F, U, R (6 variables).  
b) Modify the solve function to handle this new equation.   
c) Add a leading-zero check for T and F.                   
d) Print ALL solutions found (there may be more than one).   
e) Bonus: Count how many permutations were evaluated before   
finding the first solution.   

```python
import itertools

letters = ["T", "W", "O", "F", "U", "R"]

solutions = []
permutations_evaluated = 0
first_solution_count = None

for values in itertools.permutations(range(10), 6):

    permutations_evaluated += 1

    T, W, O, F, U, R = values

    if T == 0 or F == 0:
        continue

    TWO = 100 * T + 10 * W + O
    FOUR = 1000 * F + 100 * O + 10 * U + R

    if TWO - TWO == FOUR:

        solution = {
            "T": T,
            "W": W,
            "O": O,
            "F": F,
            "U": U,
            "R": R
        }

        solutions.append(solution)

        if first_solution_count is None:
            first_solution_count = permutations_evaluated


print("Unique Letters:")
print(letters)

print("\nTotal Solutions:", len(solutions))

if solutions:
    print("\nAll Solutions:")

    for solution in solutions:
        print(solution)

        T = solution["T"]
        W = solution["W"]
        O = solution["O"]
        F = solution["F"]
        U = solution["U"]
        R = solution["R"]

        TWO = 100 * T + 10 * W + O
        FOUR = 1000 * F + 100 * O + 10 * U + R

        print(f"{TWO} - {TWO} = {FOUR}")
else:
    print("\nNo solutions found.")

print("\nPermutations evaluated:", permutations_evaluated)

if first_solution_count is not None:
    print("Permutations before first solution:", first_solution_count)
else:
    print("No first solution exists.")
```
