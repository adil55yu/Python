Task no 2:

```python
n = int(input("How many numbers do you want to enter? "))
total = 0

for i in range(n):
    num = float(input("Enter a number: "))
    total = total + num

if n > 0:
    average = total / n
    print("Sum =", total)
    print("Average =", average)
else:
    print("Please enter a positive number of values.")
```

Task no 7:

```python
total = 0
count = 0

while True:
    num = float(input("Enter a number (0 to stop): "))

    if num == 0:
        break

    total = total + num
    count = count + 1

print("Total =", total)
print("Count =", count)

if count > 0:
    average = total / count
    print("Average =", average)
else:
    print("No non-zero numbers were entered.")
```
