Task 01 – AES Encryption using ECB Mode    
Task Statement     
Design a Python program that performs AES encryption using ECB mode. The program should take    
plaintext input from the user, apply PKCS#7 padding, and encrypt the data using a 128-bit key.    
Requirements   
1. Take plaintext input from user
2. Take 128-bit key (16 bytes)
3. Apply PKCS#7 padding
4. Encrypt using AES-ECB
5. Display results in readable format
Output Must Include  
Plaintext: ........  
Key (Hex): ........  
Padded Plaintext (Hex): ........  
Ciphertext (Hex): ........

```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad
from binascii import hexlify

plaintext = input("Enter Plaintext: ")
key_input = input("Enter 16-byte key: ")

key = key_input.encode('utf-8')
if len(key) != 16:
    print("Error: Key must be exactly 16 bytes (128-bit)")
    exit()

plaintext_bytes = plaintext.encode('utf-8')
padded_plaintext = pad(plaintext_bytes, AES.block_size)

cipher = AES.new(key, AES.MODE_ECB)
ciphertext = cipher.encrypt(padded_plaintext)

block1 = padded_plaintext[:16]
block1_encrypted = cipher.encrypt(block1)

print("\n--- Output ---")
print("Plaintext:", plaintext)
print("Key (Hex):", hexlify(key).decode())
print("Padded Plaintext (Hex):", hexlify(padded_plaintext).decode())
print("Ciphertext (Hex):", hexlify(ciphertext).decode())

print("\n--- One Block Encryption (ECB) ---")
print("Block 1 Plaintext:", hexlify(block1).decode())
print("Block 1 Ciphertext:", hexlify(block1_encrypted).decode())
```

Task 02 – AES Decryption using ECB Mode     
Task Statement   
Extend Task 01 by implementing AES decryption using ECB mode. The program should decrypt the
ciphertext and remove PKCS#7 padding to recover the original plaintext.   
Requirements
1. Take ciphertext input
2. Use same key as encryption
3. Decrypt using AES-ECB
4. Remove PKCS#7 padding manually
5. Display original plaintext   
Output Must Include   
Ciphertext (Hex): ........  
Decrypted (Padded): ........  
Padding Removed: ........  
Recovered Plaintext: ........  

```python
from Crypto.Cipher import AES
from binascii import unhexlify

ciphertext_hex = input("Enter Ciphertext (Hex): ")
key_input = input("Enter 16-byte key: ")

key = key_input.encode('utf-8')
if len(key) != 16:
    print("Error: Key must be exactly 16 bytes (128-bit)")
    exit()

ciphertext = unhexlify(ciphertext_hex)

cipher = AES.new(key, AES.MODE_ECB)
decrypted_padded = cipher.decrypt(ciphertext)

pad_len = decrypted_padded[-1]
padding_removed = decrypted_padded[:-pad_len]

print("\n--- Output ---")
print("Ciphertext (Hex):", ciphertext_hex)
print("Decrypted (Padded):", decrypted_padded.hex())
print("Padding Removed:", padding_removed.hex())
print("Recovered Plaintext:", padding_removed.decode('utf-8'))
```

Task 03 – AES Encryption using CBC Mode   
Task Statement   
Design a Python program that performs AES encryption using CBC mode. The program should use a
random Initialization Vector (IV) and demonstrate how chaining removes plaintext patterns.  
Requirements
1. Take plaintext input
2. Take 128-bit key
3. Generate random IV (16 bytes)
4. Apply PKCS#7 padding
5. Encrypt using AES-CBC   
Output Must Include  
Plaintext: ........  
Key (Hex): ........  
IV (Hex): ........  
Padded Plaintext: ........  
Ciphertext (Hex): ........  

```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad
from Crypto.Random import get_random_bytes
from binascii import hexlify

plaintext = input("Enter Plaintext: ")
key_input = input("Enter 16-byte key: ")

key = key_input.encode('utf-8')
if len(key) != 16:
    print("Error: Key must be exactly 16 bytes (128-bit)")
    exit()

iv = get_random_bytes(16)

plaintext_bytes = plaintext.encode('utf-8')
padded_plaintext = pad(plaintext_bytes, AES.block_size)

cipher = AES.new(key, AES.MODE_CBC, iv)
ciphertext = cipher.encrypt(padded_plaintext)

print("\n--- Output ---")
print("Plaintext:", plaintext)
print("Key (Hex):", hexlify(key).decode())
print("IV (Hex):", hexlify(iv).decode())
print("Padded Plaintext:", hexlify(padded_plaintext).decode())
print("Ciphertext (Hex):", hexlify(ciphertext).decode())
```

Task 04 – AES Decryption using CBC Mode   
Task Statement  
Extend Task 03 by implementing AES decryption in CBC mode. The program should correctly use the IV
and remove padding to retrieve the original plaintext.   
Requirements   
1. Input ciphertext, key, and IV   
2. Decrypt using AES-CBC   
3. Remove PKCS#7 padding     
4. Display original plaintext   
Output Must Include        
Ciphertext (Hex): ........  
IV (Hex): ........    
Decrypted (Padded): ........   
Padding Removed: ........   
Recovered Plaintext: ........  


```python
from Crypto.Cipher import AES
from binascii import unhexlify

ciphertext_hex = input("Enter Ciphertext (Hex): ")
key_input = input("Enter 16-byte key: ")
iv_hex = input("Enter IV (Hex): ")

key = key_input.encode('utf-8')
if len(key) != 16:
    print("Error: Key must be exactly 16 bytes (128-bit)")
    exit()

ciphertext = unhexlify(ciphertext_hex)
iv = unhexlify(iv_hex)

cipher = AES.new(key, AES.MODE_CBC, iv)
decrypted_padded = cipher.decrypt(ciphertext)

pad_len = decrypted_padded[-1]
padding_removed = decrypted_padded[:-pad_len]

print("\n--- Output ---")
print("Ciphertext (Hex):", ciphertext_hex)
print("IV (Hex):", iv_hex)
print("Decrypted (Padded):", decrypted_padded.hex())
print("Padding Removed:", padding_removed.hex())
print("Recovered Plaintext:", padding_removed.decode('utf-8'))
```
