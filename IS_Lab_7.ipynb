Task 1 – Caesar Brute Force Attack    
Ciphertext:     
WKH GDWD VHFXULWB GRPDLQ IRFXVHV RQ SURWHFWLQJ VHQVLWLYH     
LQIRUPDWLRQ IURP XQDXWKRULCHG DFFHVV DQG HQVXULQJ LQWHJULWB.     
Requirements:    
Try all 26 keys  
Display outputs  
Identify correct plaintext  

```python
ciphertext = "WKH GDWD VHFXULWB GRPDLQ IRFXVHV RQ SURWHFWLQJ VHQVLWLYH LQIRUPDWLRQ IURP XQDXWKRULCHG DFFHVV DQG HQVXULQJ LQWHJULWB."

def caesar_decrypt(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            shift_base = 65 if char.isupper() else 97
            result += chr((ord(char) - shift_base - shift) % 26 + shift_base)
        else:
            result += char
    return result

for key in range(26):
    decrypted_text = caesar_decrypt(ciphertext, key)
    if "THE" in decrypted_text:
        print(f"\nCorrect Key: {key}")
        print("Plaintext:", decrypted_text)
```

Task 2 – Caesar Frequency Analysis   
Ciphertext:   
WKLV PHVVDJH H[SODLQV KRZ IUHTXHQFB DQDOBVLV FDQ EH XVHG WR
EUHDN D ZHDN HQFUBSWLRQ PHWKRG.   
Requirements:  
Count frequencies  
Estimate key  
Decrypt text  

```python
ciphertext = "WKLV PHVVDJH H[SODLQV KRZ IUHTXHQFB DQDOBVLV FDQ EH XVHG WR EUHDN D ZHDN HQFUBSWLRQ PHWKRG."

def frequency_analysis(text):
    freq = {}
    for char in text:
        if char.isalpha():
            char = char.upper()
            freq[char] = freq.get(char, 0) + 1
    return freq

def caesar_decrypt(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            shift_base = 65
            result += chr((ord(char) - shift_base - shift) % 26 + shift_base)
        else:
            result += char
    return result

freq = frequency_analysis(ciphertext)

most_frequent = max(freq, key=freq.get)

key = (ord(most_frequent) - ord('E')) % 26

plaintext = caesar_decrypt(ciphertext, key)

print("Frequencies:", freq)
print("Most Frequent Letter:", most_frequent)
print("Estimated Key:", key)
print("Decrypted Text:", plaintext)
```

Task 4 – Vigenère Kasiski Attack   
Ciphertext:   
LXFOPVEFRNHRMHGLGWOGOEIBNEXMQXLXPOJYEFHNNOMHQLAZOGXHVRLXF
OPVEFRNHR   
Detect repeating sequences  
Compute distances  
Determine key length

```python
ciphertext = "LXFOPVEFRNHRMHGLGWOGOEIBNEXMQXLXPOJYEFHNNOMHQLAZOGXHVRLXFOPVEFRNHR"

seq_len = 3
sequences = {}

for i in range(len(ciphertext) - seq_len):
    seq = ciphertext[i:i+seq_len]
    if seq in sequences:
        sequences[seq].append(i)
    else:
        sequences[seq] = [i]

repeats = {k: v for k, v in sequences.items() if len(v) > 1}

distances = []
for positions in repeats.values():
    for i in range(len(positions) - 1):
        distances.append(positions[i+1] - positions[i])

print("Repeating Sequences:", repeats)
print("Distances:", distances)
```

Task 5 – Vigenère Key Recovery   
Ciphertext:   
ZICVTWQNGRZGVTWAVZHCQYGLMGJ  
Key length = 3  
Requirements:  
Split into columns
Recover key  
Perform frequency analysis   
Decrypt message


```python
ciphertext = "ZICVTWQNGRZGVTWAVZHCQYGLMGJ"
key_length = 3

def split_columns(text, key_len):
    cols = ['' for _ in range(key_len)]
    for i, char in enumerate(text):
        cols[i % key_len] += char
    return cols

def caesar_decrypt(text, shift):
    result = ""
    for char in text:
        result += chr((ord(char) - 65 - shift) % 26 + 65)
    return result

def score_text(text):
    common = ['E','T','A','O','I','N']
    return sum(text.count(c) for c in common)

columns = split_columns(ciphertext, key_length)

key = []

for col in columns:
    scores = []
    for shift in range(26):
        decrypted = caesar_decrypt(col, shift)
        scores.append((score_text(decrypted), shift))
    best_shift = max(scores)[1]
    key.append(best_shift)

key_string = ''.join(chr(k + 65) for k in key)

def decrypt_full(text, key):
    result = ""
    for i, char in enumerate(text):
        shift = key[i % len(key)]
        result += chr((ord(char) - 65 - shift) % 26 + 65)
    return result

plaintext = decrypt_full(ciphertext, key)

print("Columns:", columns)
print("Key:", key_string)
print("Plaintext:", plaintext)
```

Task 6 – Columnar Cipher Attack   
Ciphertext:  
TTNAAPTMTSUOAODWCOIXKNLYPETZHQFSECRUATYDNTOALXETM   
Requirements:   
Assume key length = 6   
Try permutations   
Recover plaintext   

```python
import itertools

ciphertext = "TTNAAPTMTSUOAODWCOIXKNLYPETZHQFSECRUATYDNTOALXETM"
key_length = 6

def split_columns_by_key(cipher, order):
    n = len(cipher)
    rows = n // key_length
    extra = n % key_length

    col_lengths = [rows + 1 if i < extra else rows for i in range(key_length)]

    cols = []
    index = 0

    for i in order:
        cols.append(cipher[index:index + col_lengths[i]])
        index += col_lengths[i]

    return cols

def read_rowwise(cols):
    text = ""
    max_len = max(len(c) for c in cols)

    for i in range(max_len):
        for col in cols:
            if i < len(col):
                text += col[i]
    return text

perms = itertools.permutations(range(key_length))

for p in perms:
    cols = split_columns_by_key(ciphertext, p)
    pt = read_rowwise(cols)

    if "THE" in pt or "AND" in pt or "SECRET" in pt:
        print("Key Order:", p)
        print("Plaintext:", pt)
        break
```
