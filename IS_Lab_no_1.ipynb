Develop a program that:

3. Determines whether the number is Even or Odd

Task 4 — Simple Calculator
Design a calculator program that:
1. Takes two numbers as input
2. Takes an operator (+, −, ×, ÷)
3. Performs the selected operation using conditional statements

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero")
else:
    print("Invalid operator")
```

Task 8 — Prime Number Checker
Create a program that:
1. Takes an integer input
2. Checks whether the number is prime or not


```python
num = int(input("Enter an integer: "))

if num <= 1:
    print(f"{num} is not a prime number")
else:
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is not a prime number")
```

Task 9 — Number Guessing Simulation
Design a program that:
1. Stores a predefined number
2. Allows the user to guess the number
3. Repeats until the correct guess is entered
4. Displays the total number of attempts

```python
secret_number = 42
attempts = 0
guess = None

while guess != secret_number:
    guess = int(input("Guess the number: "))
    attempts += 1
    if guess < secret_number:
        print("Too low! Try again.")
    elif guess > secret_number:
        print("Too high! Try again.")

print(f"Congratulations! You guessed the number {secret_number} in {attempts} attempts.")
```

Task 10 — Password Retry System
Write a program that:
1. Stores a predefined password
2. Allows a maximum of three attempts
3. Displays:
• “Access Granted” if correct
• “Account Locked” after three failed attempts

```python
password = "secure123"
attempts = 0
max_attempts = 3

while attempts < max_attempts:
    user_input = input("Enter password: ")
    if user_input == password:
        print("Access Granted")
        break
    else:
        attempts += 1
        if attempts < max_attempts:
            print("Incorrect password. Try again.")
        else:
            print("Account Locked")
```

Task 12 — Reverse Number Program
Develop a program that:
1. Takes an integer input
2. Reverses the digits of the number

```python
num = int(input("Enter an integer: "))
reversed_num = 0
temp = abs(num)

while temp > 0:
    digit = temp % 10
    reversed_num = reversed_num * 10 + digit
    temp //= 10

if num < 0:
    reversed_num = -reversed_num

print(f"Reversed Number: {reversed_num}")
```

Task 13 — Digit Counter System
Write a program that:
1. Takes an integer input
2. Counts and displays the total number of digits

```python
num = int(input("Enter an integer: "))
temp = abs(num)
count = 0

if temp == 0:
    count = 1
else:
    while temp > 0:
        temp //= 10
        count += 1

print(f"Total number of digits: {count}")
```

Task 14 — Even/Odd Counter
Create a program that:
1. Takes a number n as input
2. Counts how many even and odd numbers exist from 1 to n

```python
n = int(input("Enter a number: "))
even_count = 0
odd_count = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print(f"Total even numbers from 1 to {n}: {even_count}")
print(f"Total odd numbers from 1 to {n}: {odd_count}")
```
