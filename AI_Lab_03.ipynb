1. Make a class parallelogram and write a function to calculate its area.

```python
class Parallelogram:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def calculate_area(self):
        return self.base * self.height


p1 = Parallelogram(12, 8)

print("Base:", p1.base)
print("Height:", p1.height)
print("Area of Parallelogram:", p1.calculate_area())
```

2. You are working on a Python program to simulate a car dealership&#39;s inventory
management system. The system aims to model cars and their attributes
accurately.         
 You are tasked with creating a Python program to represent vehicles using a
class named Car.     
 Use the __init__() constructor to initialize:   
 car brand   
 model   
 price   
 color      
 Create at least one function (method) inside the class to display car details.     
 Create multiple objects of the class representing different cars.    
 Call the class function using objects to display the information of each car.

```python
class Car:
    def __init__(self, brand, model, price, color):
        self.brand = brand
        self.model = model
        self.price = price
        self.color = color

    def display(self):
        print("Brand :", self.brand)
        print("Model :", self.model)
        print("Price :", self.price)
        print("Color :", self.color)
        print("---------------------------")


car1 = Car("Toyota", "Corolla", 5500000, "White")
car2 = Car("Honda", "Civic", 6200000, "Black")
car3 = Car("Suzuki", "Swift", 4200000, "Red")

car1.display()
car2.display()
car3.display()
```
