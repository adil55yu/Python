Task 1 — User Database Management System   
Design a system that:   
• Stores 5 usernames in a list   
• Displays all usernames   
• Sorts usernames in ascending order   
• Sorts usernames in descending order   
• Searches for a specific username  
• Adds a new username  
• Removes an existing username  

```python
usernames = ["Ali", "Ahmed", "Sara", "Zain", "Hina"]

while True:
    print("\n--- User Database Management System ---")
    print("1. Display all usernames")
    print("2. Sort usernames (Ascending)")
    print("3. Sort usernames (Descending)")
    print("4. Search for a username")
    print("5. Add a new username")
    print("6. Remove an existing username")
    print("7. Exit")

    choice = input("Enter your choice (1-7): ")

    if choice == "1":
        print("Usernames:", usernames)

    elif choice == "2":
        usernames.sort()
        print("Usernames sorted in ascending order:", usernames)

    elif choice == "3":
        usernames.sort(reverse=True)
        print("Usernames sorted in descending order:", usernames)

    elif choice == "4":
        search_name = input("Enter username to search: ")
        if search_name in usernames:
            print(search_name, "found in database.")
        else:
            print(search_name, "not found.")

    elif choice == "5":
        new_user = input("Enter new username: ")
        usernames.append(new_user)
        print(new_user, "added successfully.")

    elif choice == "6":
        remove_user = input("Enter username to remove: ")
        if remove_user in usernames:
            usernames.remove(remove_user)
            print(remove_user, "removed successfully.")
        else:
            print("Username not found.")

    elif choice == "7":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Please try again.")
```

Task 3 — Encryption Key Sorting System   
Develop a program that:  
• Generates 10 random keys (range 1000–9999)  
• Stores keys in a list  
• Sorts keys in ascending order  
• Sorts keys in descending order  
• Displays strongest and weakest key  

```python
import random

keys = []

for i in range(10):
    key = random.randint(1000, 9999)
    keys.append(key)

print("Generated Keys:", keys)

keys.sort()
print("Keys in Ascending Order:", keys)

keys.sort(reverse=True)
print("Keys in Descending Order:", keys)

strongest_key = max(keys)
weakest_key = min(keys)

print("Strongest Key:", strongest_key)
print("Weakest Key:", weakest_key)
```

Task 4 — Columnar Key Ordering System  
Program should:   
• Take cipher key input  
• Convert key to list  
• Sort alphabetically  
• Display sorted key  
• Display original index mapping  


```python
key = input("Enter cipher key: ")

key_list = list(key)

sorted_key = sorted(key_list)

print("Sorted Key:", "".join(sorted_key))

index_mapping = []

for char in sorted_key:
    index = key_list.index(char)
    index_mapping.append(index)
    key_list[index] = None

print("Original Index Mapping:", index_mapping)
```

Task 5 — Cipher Grid Generator  
Design a system that:  
• Takes plaintext input  
• Takes number of columns  
• Stores plaintext in 2D grid row-wise    
• Displays formatted grid  

```python
plaintext = input("Enter plaintext: ")
columns = int(input("Enter number of columns: "))

rows = len(plaintext)
if len(plaintext) % columns != 0:
    rows += 1

grid = []
index = 0

for r in range(rows):
    row = []
    for c in range(columns):
        if index < len(plaintext):
            row.append(plaintext[index])
        else:
            row.append(" ")
        index += 1
    grid.append(row)

print("\nFormatted Grid:")
for row in grid:
    print(" ".join(row))
```

Task 6 — Random Password Repository   
Create a system that:  
• Generates 5 random passwords  
• Each password length = 8  
• Stores passwords in list    
• Sorts passwords alphabetically   
• Displays sorted list

```python
import random
import string

passwords = []

for i in range(5):
    password = ""
    for j in range(8):
        password += random.choice(string.ascii_letters + string.digits)
    passwords.append(password)

passwords.sort()

print("Sorted Passwords:")
for pwd in passwords:
    print(pwd)
```

Task 7 — Directory Explorer Using OS Library    
Program should:   
• Display current working directory   
• List all files   
• Count total files   
• Display only .txt files   

```python
import os

current_directory = os.getcwd()
print("Current Working Directory:", current_directory)

items = os.listdir(current_directory)

files = []
txt_files = []

for item in items:
    if os.path.isfile(item):
        files.append(item)
        if item.endswith(".txt"):
            txt_files.append(item)

print("\nAll Files:")
for file in files:
    print(file)

print("\nTotal Files:", len(files))

print("\nText Files (.txt only):")
for txt in txt_files:
    print(txt)
```

Task 8 — OTP Verification System   
Design a system that:   
• Generates 6-digit OTP   
• Stores OTP                
• Asks user to re-enter OTP    
• Verifies correctness   
• Allows 3 attempts     


```python
import random

otp = random.randint(100000, 999999)

print("Your OTP is:", otp)

attempts = 3

while attempts > 0:
    user_otp = int(input("Enter the OTP: "))

    if user_otp == otp:
        print("OTP Verified Successfully!")
        break
    else:
        attempts -= 1
        if attempts > 0:
            print("Incorrect OTP. Attempts left:", attempts)
        else:
            print("Verification Failed. No attempts left.")
```
