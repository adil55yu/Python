Lab Task 1: Encrypt a message with Caesar Cipher
Write a program that:
1. Asks the user for a message and a shift value.
2. Encrypts the message using Caesar Cipher.
3. Prints the encrypted message.

Write a program that:
1. Asks the user for a message and a ciphertext.
2. Decrypts the message using Caesar Cipher.
3. Prints the Decrypted message.

```python
message = input("Enter message: ")
shift = int(input("Enter shift value: "))

encrypted_message = ""

for char in message:
    if char.isalpha():
        ascii_offset = 65 if char.isupper() else 97
        encrypted_char = chr((ord(char) - ascii_offset + shift) % 26 + ascii_offset)
        encrypted_message += encrypted_char
    else:
        encrypted_message += char

print("Encrypted Message:", encrypted_message)

decrypted_message = ""

for char in encrypted_message:
    if char.isalpha():
        ascii_offset = 65 if char.isupper() else 97
        decrypted_char = chr((ord(char) - ascii_offset - shift) % 26 + ascii_offset)
        decrypted_message += decrypted_char
    else:
        decrypted_message += char

print("Decrypted Message:", decrypted_message)
```
