Task 1: Write a python program to implement DFS and BFS on the graph given below. Consider
this to be a directed graph. Start state =1 and Goal state = 11. Your program should stop once Goal
state is reached.

```python
from collections import deque

graph = {
    1: [2, 3, 4],
    2: [5, 6],
    3: [],
    4: [7, 8],
    5: [9, 10],
    6: [],
    7: [11],
    8: [12],
    9: [],
    10: [],
    11: [],
    12: []
}

def bfs(graph, start, goal):
    queue = deque([start])
    visited = set([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        if node == goal:
            print("Goal Found!")
            return order

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order


def dfs(graph, start, goal):
    stack = [start]
    visited = set()
    order = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            order.append(node)

            if node == goal:
                print("Goal Found!")
                return order

            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return order


start = 1
goal = 11

print("Breadth First Search")
bfs_order = bfs(graph, start, goal)
print("Traversal:", bfs_order)

print("\nDepth First Search")
dfs_order = dfs(graph, start, goal)
print("Traversal:", dfs_order)
```

Task 2: Write a python program to implement DFS and BFS on the graph given below. Input
should be taken from the user.    
Start state: A,   
Goal state: G.                  
When the program reaches the Goal state, the traversal should stop and prints the path.

```python
from collections import deque

graph = {
    'A': ['D', 'B'],
    'B': ['C', 'E', 'G'],
    'C': ['A'],
    'D': ['C'],
    'E': ['H'],
    'F': [],
    'G': ['F'],
    'H': ['G', 'F']
}

def bfs(graph, start, goal):
    queue = deque([(start, [start])])
    visited = set([start])

    while queue:
        node, path = queue.popleft()

        if node == goal:
            print("\nGoal Found!")
            print("Path:", " -> ".join(path))
            return

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    print("Goal not found.")


def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()

    while stack:
        node, path = stack.pop()

        if node not in visited:
            visited.add(node)

            if node == goal:
                print("\nGoal Found!")
                print("Path:", " -> ".join(path))
                return

            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor]))

    print("Goal not found.")


start = input("Enter Start Node: ").upper()
goal = input("Enter Goal Node: ").upper()

print("\nBreadth First Search")
bfs(graph, start, goal)

print("\nDepth First Search")
dfs(graph, start, goal)
```

```python
from collections import deque

graph = {
    'A': [('B', 5), ('C', 7)],
    'B': [('D', 10), ('E', 10)],
    'C': [('F', 8)],
    'D': [],
    'E': [],
    'F': []
}

def bfs(start, goal):
    queue = deque([(start, 0)])   # (node, cumulative_cost)
    visited = set()
    expansion_order = []

    while queue:
        node, cost = queue.popleft()

        if node not in visited:
            visited.add(node)
            expansion_order.append(node)

            if node == goal:
                print("Expansion Order:", expansion_order)
                print("Total Path Cost:", cost)
                return

            for neighbor, edge_cost in graph[node]:
                if neighbor not in visited:
                    queue.append((neighbor, cost + edge_cost))

goal = "F"
bfs("A", goal)
```
