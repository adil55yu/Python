Task 1 – Vigenère Cipher Encryption Program
Design a Python program that:
1. Takes plaintext input from the user
2. Takes a keyword as the encryption key
3. Repeats the key so that its length matches the plaintext
4. Converts plaintext and key letters into alphabet positions
5. Applies the Vigenère encryption formula
6. Generates the final ciphertext
7. Displays plaintext, key and ciphertext


```python
plaintext = input("Enter Plaintext: ")
key = input("Enter Key: ")

key_repeated = ""
key_index = 0

for char in plaintext:
    if char.isalpha():
        key_repeated += key[key_index % len(key)]
        key_index += 1
    else:
        key_repeated += char

ciphertext = ""

for p, k in zip(plaintext, key_repeated):
    if p.isalpha():

        if p.isupper():
            base = 65
        else:
            base = 97

        p_val = ord(p) - base
        k_val = ord(k.lower()) - 97

        c_val = (p_val + k_val) % 26
        ciphertext += chr(c_val + base)

    else:
        ciphertext += p

print("Plaintext :", plaintext)
print("Key       :", key_repeated)
print("Ciphertext:", ciphertext)
```

Task 2 – Vigenère Cipher Decryption Program
Design a Python program that:
1. Takes ciphertext as input
2. Takes the encryption key
3. Repeats the key to match ciphertext length
4. Converts letters into alphabet positions
5. Applies the Vigenère decryption formula
6. Recovers the original plaintext
7. Displays ciphertext, key and decrypted plaintext


```python
ciphertext = input("Enter Ciphertext: ")
key = input("Enter Key: ")

key_repeated = ""
key_index = 0

for char in ciphertext:
    if char.isalpha():
        key_repeated += key[key_index % len(key)]
        key_index += 1
    else:
        key_repeated += char

plaintext = ""

for c, k in zip(ciphertext, key_repeated):
    if c.isalpha():

        if c.isupper():
            base = 65
        else:
            base = 97

        c_val = ord(c) - base
        k_val = ord(k.lower()) - 97

        p_val = (c_val - k_val) % 26
        plaintext += chr(p_val + base)

    else:
        plaintext += c

print("Ciphertext :", ciphertext)
print("Key        :", key_repeated)
print("Plaintext  :", plaintext)
```

Task 3 – Complete Vigenère Cipher System
Design a complete Python program that allows the user to choose between:
1. Encryption
2. Decryption
Program requirements:
1. Display a menu to the user
2. Allow the user to select encryption or decryption
3. Accept message input
4. Accept keyword input
5. Perform the selected operation
6. Display the final result


```python
print("Vigenere Cipher System")
print("1. Encrypt Message")
print("2. Decrypt Message")

choice = input("Enter your choice (1 or 2): ")
message = input("Enter the message: ")
key = input("Enter the key: ")

key_repeated = ""
key_index = 0

for char in message:
    if char.isalpha():
        key_repeated += key[key_index % len(key)]
        key_index += 1
    else:
        key_repeated += char

result = ""

for m, k in zip(message, key_repeated):
    if m.isalpha():

        if m.isupper():
            base = 65
        else:
            base = 97

        m_val = ord(m) - base
        k_val = ord(k.lower()) - 97

        if choice == "1":
            value = (m_val + k_val) % 26
        elif choice == "2":
            value = (m_val - k_val) % 26
        else:
            print("Invalid choice")
            break

        result += chr(value + base)

    else:
        result += m

print("Message :", message)
print("Key     :", key_repeated)

if choice == "1":
    print("Encrypted Message :", result)
elif choice == "2":
    print("Decrypted Message :", result)
```
