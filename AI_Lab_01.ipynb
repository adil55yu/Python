Task 1:
Write a Python program to find those numbers which are divisible by 7 and
multiple of 5, between 1500 and 2700 (both included)

```python
for i in range(1500, 2701):
    if i % 7 == 0 and i % 5 == 0:
        print(i, end=" ")
```

Task 2:
Write a Python program to find numbers between 100 and 400 (both included)
where each digit of a number is an even number. The numbers obtained should be
printed in a comma-separated sequence.

```python
result = []

for num in range(100, 401):
    num_str = str(num)

    if (int(num_str[0]) % 2 == 0 and
        int(num_str[1]) % 2 == 0 and
        int(num_str[2]) % 2 == 0):
        result.append(num_str)

print(",".join(result))
```

Task 3:
Write a Python program to calculate the sum and average of n integer numbers
(input from the user). Input 0 to finish.

```python
sum = 0
count = 0

while True:
    num = int(input("Enter a number (0 to finish): "))

    if num == 0:
        break

    sum = sum + num
    count = count + 1

if count > 0:
    average = sum / count
    print("Sum =", sum)
    print("Average =", average)
else:
    print("No numbers were entered.")
```

Task 4:   
A student will not be allowed to sit in exam if his/her attendance is less than 75%.                              
Take following input from user      
 Number of classes held         
 Number of classes attended.            
 And print percentage of class attended          
 Is student is allowed to sit in exam or not.

```python
classes_held = int(input("Enter the number of classes held: "))
classes_attended = int(input("Enter the number of classes attended: "))

percentage = (classes_attended / classes_held) * 100

print("Attendance Percentage =", percentage, "%")

if percentage >= 75:
    print("Student is allowed to sit in the exam.")
else:
    print("Student is NOT allowed to sit in the exam.")
```

Task 5:   
Write a Python program to perform the following tasks using lists:      
 Create a list named numbers containing integers from 1 to 5.    
 Print the original list.         
 Append the number 6 to the list.                   
 Insert the number 0 at the beginning of the list.       
 Print the modified list.           
 Remove the number 3 from the list.       
 Print the final list.          

```python
numbers = [1, 2, 3, 4, 5]

print("Original List:", numbers)

numbers.append(6)

numbers.insert(0, 0)

print("Modified List:", numbers)

numbers.remove(3)

print("Final List:", numbers)
```
