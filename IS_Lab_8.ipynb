Task 1 – DES Encryption & Decryption System   
Design a Python program that:   
Input:   
Take input from user and convert in blocks of 64-bit plaintext (binary or hex)  
Take input from user and convert in block of 64-bit key   
Note apply case sensitivity for both.   
Requirements:               
1. Apply Initial Permutation   
2. Split into L and R                  
3. Implement at least 2 rounds of DES:    
i. Expansion (E-box)   
ii. XOR with key   
iii. S-box substitution  
iv. P-box permutation   
v. XOR with left half   
vi. Swap   
4. Implement Final Permutation   
5. Display:   
i. Intermediate values (each step)   
ii. Final ciphertext   
6. Implement decryption:   
i. Use same logic   
ii. Reverse keys   
Output must include:   
After IP: ........   
After Round 1: ........   
After Round 2: ........   
Ciphertext: ........  
Decrypted Text: ........   

```python
IP = [58,50,42,34,26,18,10,2,
      60,52,44,36,28,20,12,4,
      62,54,46,38,30,22,14,6,
      64,56,48,40,32,24,16,8,
      57,49,41,33,25,17,9,1,
      59,51,43,35,27,19,11,3,
      61,53,45,37,29,21,13,5,
      63,55,47,39,31,23,15,7]

FP = [40,8,48,16,56,24,64,32,
      39,7,47,15,55,23,63,31,
      38,6,46,14,54,22,62,30,
      37,5,45,13,53,21,61,29,
      36,4,44,12,52,20,60,28,
      35,3,43,11,51,19,59,27,
      34,2,42,10,50,18,58,26,
      33,1,41,9,49,17,57,25]

E = [32,1,2,3,4,5,
     4,5,6,7,8,9,
     8,9,10,11,12,13,
     12,13,14,15,16,17,
     16,17,18,19,20,21,
     20,21,22,23,24,25,
     24,25,26,27,28,29,
     28,29,30,31,32,1]

P = [16,7,20,21,29,12,28,17,
     1,15,23,26,5,18,31,10,
     2,8,24,14,32,27,3,9,
     19,13,30,6,22,11,4,25]

SBOX = [
    [14,4,13,1,2,15,11,8,3,10,6,12,5,9,0,7],
    [0,15,7,4,14,2,13,1,10,6,12,11,9,5,3,8],
    [4,1,14,8,13,6,2,11,15,12,9,7,3,10,5,0],
    [15,12,8,2,4,9,1,7,5,11,3,14,10,0,6,13]
]

def permute(block, table):
    return ''.join(block[i-1] for i in table)

def xor(a, b):
    return ''.join('0' if a[i] == b[i] else '1' for i in range(len(a)))

def sbox_substitution(bits):
    output = ""
    for i in range(0, len(bits), 6):
        block = bits[i:i+6]
        row = int(block[0] + block[5], 2)
        col = int(block[1:5], 2)
        val = SBOX[row][col]
        output += format(val, '04b')
    return output

def to_binary(text):
    if all(c in '01' for c in text):
        return text.zfill(64)
    else:
        return bin(int(text, 16))[2:].zfill(64)

def des_round(L, R, key):
    expanded = permute(R, E)
    print("Expansion:", expanded)

    xored = xor(expanded, key)
    print("XOR with Key:", xored)

    sbox_out = sbox_substitution(xored)
    print("S-box:", sbox_out)

    pbox_out = permute(sbox_out, P)
    print("P-box:", pbox_out)

    new_R = xor(L, pbox_out)
    new_L = R

    return new_L, new_R

def des_encrypt(plaintext, key):
    plaintext = to_binary(plaintext)
    key = to_binary(key)

    print("\nPlaintext:", plaintext)
    print("Key:", key)

    ip = permute(plaintext, IP)
    print("\nAfter IP:", ip)

    L, R = ip[:32], ip[32:]

    K1 = key[:48]
    K2 = key[16:64]

    print("\n--- Round 1 ---")
    L, R = des_round(L, R, K1)
    print("After Round 1:", L + R)

    print("\n--- Round 2 ---")
    L, R = des_round(L, R, K2)
    print("After Round 2:", L + R)

    combined = R + L
    ciphertext = permute(combined, FP)

    print("\nCiphertext:", ciphertext)
    return ciphertext, [K1, K2]

def des_decrypt(ciphertext, keys):
    ip = permute(ciphertext, IP)
    L, R = ip[:32], ip[32:]

    print("\n--- Decryption ---")

    for i, key in enumerate(reversed(keys)):
        print(f"\n--- Round {i+1} ---")
        L, R = des_round(L, R, key)

    combined = R + L
    plaintext = permute(combined, FP)

    print("\nDecrypted Text:", plaintext)
    return plaintext

plaintext = input("Enter Plaintext (Hex/Binary): ")
key = input("Enter Key (Hex/Binary): ")

cipher, keys = des_encrypt(plaintext, key)
des_decrypt(cipher, keys)
```

Task 2 – S-DES Based MITM Attack   
Design a Python program that:  
Step 1 – Double Encryption  
C = E(K2, E(K1, P))   
Step 2 – MITM Attack  
Program must:                     
1. Generate all possible keys K1    
2. Encrypt plaintext with each K1   
3. Store results in dictionary   
4. Generate all possible K2          
5. Decrypt ciphertext with each K2   
6. Match with stored values   
Output must include:   
Display:           
Recovered K1 = ...   
Recovered K2 = ...       
Recovered Plaintext = ...   

```python
P10 = [3,5,2,7,4,10,1,9,8,6]
P8 = [6,3,7,4,8,5,10,9]
P4 = [2,4,3,1]
IP = [2,6,3,1,4,8,5,7]
IP_INV = [4,1,3,5,7,2,8,6]
EP = [4,1,2,3,2,3,4,1]

S0 = [
    [1,0,3,2],
    [3,2,1,0],
    [0,2,1,3],
    [3,1,3,2]
]

S1 = [
    [0,1,2,3],
    [2,0,1,3],
    [3,0,1,0],
    [2,1,0,3]
]

def permute(bits, table):
    return ''.join(bits[i-1] for i in table)

def left_shift(bits):
    return bits[1:] + bits[0]

def xor(a, b):
    return ''.join('0' if a[i] == b[i] else '1' for i in range(len(a)))

def sbox(bits, box):
    row = int(bits[0] + bits[3], 2)
    col = int(bits[1] + bits[2], 2)
    return format(box[row][col], '02b')

def generate_keys(key):
    key = permute(key, P10)
    left, right = key[:5], key[5:]

    left = left_shift(left)
    right = left_shift(right)
    k1 = permute(left + right, P8)

    left = left_shift(left)
    right = left_shift(right)
    k2 = permute(left + right, P8)

    return k1, k2

def fk(bits, key):
    left, right = bits[:4], bits[4:]
    temp = permute(right, EP)
    temp = xor(temp, key)

    left_s = sbox(temp[:4], S0)
    right_s = sbox(temp[4:], S1)

    combined = permute(left_s + right_s, P4)
    left = xor(left, combined)

    return left + right

def sdes_encrypt(plain, key):
    k1, k2 = generate_keys(key)
    bits = permute(plain, IP)
    bits = fk(bits, k1)
    bits = bits[4:] + bits[:4]
    bits = fk(bits, k2)
    return permute(bits, IP_INV)

def sdes_decrypt(cipher, key):
    k1, k2 = generate_keys(key)
    bits = permute(cipher, IP)
    bits = fk(bits, k2)
    bits = bits[4:] + bits[:4]
    bits = fk(bits, k1)
    return permute(bits, IP_INV)

def double_encrypt(p, k1, k2):
    return sdes_encrypt(sdes_encrypt(p, k1), k2)

def mitm_attack(p, c):
    table = {}

    for i in range(1024):
        k1 = format(i, '010b')
        mid = sdes_encrypt(p, k1)
        table[mid] = k1

    for i in range(1024):
        k2 = format(i, '010b')
        mid = sdes_decrypt(c, k2)

        if mid in table:
            return table[mid], k2

    return None, None

p = input("Enter 8-bit Plaintext: ")
k1_real = input("Enter real K1 (10-bit): ")
k2_real = input("Enter real K2 (10-bit): ")

c = double_encrypt(p, k1_real, k2_real)

print("Ciphertext:", c)

k1_found, k2_found = mitm_attack(p, c)

print("Recovered K1 =", k1_found)
print("Recovered K2 =", k2_found)
print("Recovered Plaintext =", sdes_decrypt(sdes_decrypt(c, k2_found), k1_found))
```
