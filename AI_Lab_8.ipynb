Task 1: Write a Python program to implement the AC-3 algorithm for the given
Australia map coloring problem. Your program should:                       
 Define variables, domains, and constraints  
 Implement the REVISE function   
 Implement the AC-3 algorithm                  
 Display domain values after each revision step  
 Print the final reduced domains   

Focus: Show how domains shrink step-by-step ... not just final output

```python
from collections import deque

variables = [
    "WA", "NT", "SA",
    "Q", "NSW", "V", "T"
]


domains = {
    "WA":  ["Red", "Green", "Blue"],
    "NT":  ["Red", "Green", "Blue"],
    "SA":  ["Red", "Green", "Blue"],
    "Q":   ["Red", "Green", "Blue"],
    "NSW": ["Red", "Green", "Blue"],
    "V":   ["Red", "Green", "Blue"],
    "T":   ["Red", "Green", "Blue"]
}


neighbors = {
    "WA":  ["NT", "SA"],
    "NT":  ["WA", "SA", "Q"],
    "SA":  ["WA", "NT", "Q", "NSW", "V"],
    "Q":   ["NT", "SA", "NSW"],
    "NSW": ["SA", "Q", "V"],
    "V":   ["SA", "NSW"],
    "T":   []
}


def print_domains(domains):
    for variable in variables:
        print(f"  {variable}: {domains[variable]}")
    print()




def revise(xi, xj, domains):
    """
    REVISE removes values from Xi that have no
    compatible value in Xj.

    Returns:
        True  -> if Xi's domain was changed
        False -> if Xi's domain was not changed
    """

    revised = False

    for value_x in domains[xi][:]:

        supported = False

        for value_y in domains[xj]:

            if value_x != value_y:
                supported = True
                break

        if not supported:
            domains[xi].remove(value_x)
            revised = True

    return revised


def ac3(domains):

    queue = deque()

    for xi in variables:
        for xj in neighbors[xi]:
            queue.append((xi, xj))

    step = 1

    print("=" * 60)
    print("INITIAL DOMAINS")
    print("=" * 60)
    print_domains(domains)

    while queue:

        xi, xj = queue.popleft()

        print("-" * 60)
        print(f"STEP {step}: REVISE({xi}, {xj})")
        print("-" * 60)

        old_domain = domains[xi].copy()

        if revise(xi, xj, domains):

            print(f"Domain of {xi} changed:")
            print(f"Before: {old_domain}")
            print(f"After : {domains[xi]}")
            print()

            if len(domains[xi]) == 0:
                print(f"Domain of {xi} is empty!")
                return False

            for xk in neighbors[xi]:
                if xk != xj:
                    queue.append((xk, xi))

        else:
            print(f"No change in domain of {xi}.")
            print(f"Domain: {domains[xi]}")
            print()

        print("Current domains:")
        print_domains(domains)

        step += 1

    return True


print("=" * 60)
print("AUSTRALIA MAP COLORING USING AC-3")
print("=" * 60)

result = ac3(domains)


print("=" * 60)
print("FINAL REDUCED DOMAINS")
print("=" * 60)

print_domains(domains)

if result:
    print("AC-3 completed successfully.")
else:
    print("CSP has no solution.")
```

Task 2: Extend your AC-3 program to assign final colors to all regions of map using
Backtracking Search after applying AC-3. Your program should:    
 Take reduced domains from AC-3                          
 Assign values such that all constraints are satisfied       
 Print the final valid coloring of the map    

```python
from collections import deque

variables = [
    "WA", "NT", "SA",
    "Q", "NSW", "V", "T"
]

domains = {
    "WA":  ["Red", "Green", "Blue"],
    "NT":  ["Red", "Green", "Blue"],
    "SA":  ["Red"],
    "Q":   ["Red", "Green", "Blue"],
    "NSW": ["Red", "Green", "Blue"],
    "V":   ["Red", "Green", "Blue"],
    "T":   ["Red", "Green", "Blue"]
}

neighbors = {
    "WA":  ["NT", "SA"],
    "NT":  ["WA", "SA", "Q"],
    "SA":  ["WA", "NT", "Q", "NSW", "V"],
    "Q":   ["NT", "SA", "NSW"],
    "NSW": ["SA", "Q", "V"],
    "V":   ["SA", "NSW"],
    "T":   []
}

def print_domains(domains):
    for variable in variables:
        print(f"{variable}: {domains[variable]}")
    print()

def revise(xi, xj, domains):
    revised = False

    for value_x in domains[xi][:]:
        supported = False

        for value_y in domains[xj]:
            if value_x != value_y:
                supported = True
                break

        if not supported:
            domains[xi].remove(value_x)
            revised = True

    return revised

def ac3(domains):
    queue = deque()

    for xi in variables:
        for xj in neighbors[xi]:
            queue.append((xi, xj))

    step = 1

    print("=" * 60)
    print("INITIAL DOMAINS")
    print("=" * 60)
    print_domains(domains)

    while queue:
        xi, xj = queue.popleft()
        old_domain = domains[xi].copy()

        if revise(xi, xj, domains):
            print("-" * 60)
            print(f"STEP {step}: REVISE({xi}, {xj})")
            print("-" * 60)
            print(f"{xi} before: {old_domain}")
            print(f"{xi} after : {domains[xi]}")

            print("\nCurrent Domains:")
            print_domains(domains)

            if len(domains[xi]) == 0:
                return False

            for xk in neighbors[xi]:
                if xk != xj:
                    queue.append((xk, xi))

            step += 1

    return True

def is_consistent(variable, color, assignment):
    for neighbor in neighbors[variable]:
        if neighbor in assignment:
            if assignment[neighbor] == color:
                return False

    return True

def backtracking_search(assignment, domains):
    if len(assignment) == len(variables):
        return assignment

    unassigned = [
        variable
        for variable in variables
        if variable not in assignment
    ]

    variable = min(
        unassigned,
        key=lambda v: len(domains[v])
    )

    print(f"\nTrying to assign {variable}")
    print(f"Available colors: {domains[variable]}")

    for color in domains[variable]:
        print(f"Trying {variable} = {color}")

        if is_consistent(variable, color, assignment):
            assignment[variable] = color

            print("Assignment accepted.")
            print("Current assignment:", assignment)

            result = backtracking_search(
                assignment,
                domains
            )

            if result is not None:
                return result

            print(f"Backtracking from {variable} = {color}")
            del assignment[variable]

        else:
            print("Conflict! Color cannot be used.")

    return None

print("=" * 60)
print("AUSTRALIA MAP COLORING")
print("AC-3 + BACKTRACKING SEARCH")
print("=" * 60)

print("\n" + "=" * 60)
print("STEP 1: APPLYING AC-3")
print("=" * 60)

if ac3(domains):

    print("\nAC-3 completed successfully.")
    print("\nReduced Domains:")
    print_domains(domains)

else:
    print("AC-3 failed. No solution exists.")
    exit()

print("\n" + "=" * 60)
print("STEP 2: BACKTRACKING SEARCH")
print("=" * 60)

assignment = {}

solution = backtracking_search(
    assignment,
    domains
)

print("\n" + "=" * 60)
print("FINAL VALID COLORING")
print("=" * 60)

if solution:
    for variable in variables:
        print(f"{variable} = {solution[variable]}")
else:
    print("No valid coloring found.")
```
