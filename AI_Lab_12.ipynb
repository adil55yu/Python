```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

import warnings
warnings.filterwarnings('ignore')

# Load Dataset
df = pd.read_csv('/content/customer_segmentation2.csv')

print("First 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# --- Data Preprocessing for New Dataset ---
# Handle missing 'Income' values
df['Income'].fillna(df['Income'].median(), inplace=True)

# Create 'Age' feature from 'Year_Birth'
# Assuming current year is 2023 for age calculation
df['Age'] = 2023 - df['Year_Birth']

# Create 'Annual Income (k$)' from 'Income'
df['Annual Income (k$)'] = df['Income'] / 1000

# Create 'Spending Score (1-100)' by summing up all 'Mnt' columns
mnt_cols = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]
df['Spending Score (1-100)'] = df[mnt_cols].sum(axis=1)

# --- Original Analysis (adapted for new features) ---

# Gender Distribution (Removed as 'Gender' column is not available)
# plt.figure(figsize=(7,5))
# sns.countplot(x='Gender', data=df)
# plt.title('Customer Distribution by Gender')
# plt.xlabel('Gender')
# plt.ylabel('Number of Customers')
# plt.show()

# Age Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution of Customers')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.show()

# Annual Income Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Annual Income (k$)'], bins=20, kde=True)
plt.title('Annual Income Distribution')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Number of Customers')
plt.show()

# Spending Score Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Spending Score (1-100)'], bins=20, kde=True)
plt.title('Spending Score Distribution')
plt.xlabel('Spending Score')
plt.ylabel('Number of Customers')
plt.show()

# Age Boxplot
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Age'])
plt.title('Age Boxplot')
plt.xlabel('Age')
plt.show()

# Annual Income Boxplot
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Annual Income (k$)'])
plt.title('Annual Income Boxplot')
plt.xlabel('Annual Income (k$)')
plt.show()

# Spending Score Boxplot
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Spending Score (1-100)'])
plt.title('Spending Score Boxplot')
plt.xlabel('Spending Score')
plt.show()

# Age vs Annual Income
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Age',
    y='Annual Income (k$)',
    data=df,
    s=80
)
plt.title('Age vs Annual Income')
plt.xlabel('Age')
plt.ylabel('Annual Income (k$)')
plt.show()

# Age vs Spending Score
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Age',
    y='Spending Score (1-100)',
    data=df,
    s=80
)
plt.title('Age vs Spending Score')
plt.xlabel('Age')
plt.ylabel('Spending Score')
plt.show()

# Annual Income vs Spending Score
plt.figure(figsize=(9,6))
sns.scatterplot(
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    data=df,
    s=100
)
plt.title('Annual Income vs Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()

# Correlation Heatmap
numeric_columns = [
    'Age',
    'Annual Income (k$)',
    'Spending Score (1-100)'
]

plt.figure(figsize=(8,6))
correlation = df[numeric_columns].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()

# Pairplot (Removed 'hue="Gender"' as 'Gender' column is not available)
sns.pairplot(
    df,
    vars=[
        'Age',
        'Annual Income (k$)',
        'Spending Score (1-100)'
    ]
)
plt.show()

# Select Features for K-Means
X = df[
    [
        'Annual Income (k$)',
        'Spending Score (1-100)'
    ]
]

# Elbow Method
wcss = []

for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i,
        init='k-means++',
        random_state=42,
        n_init=10
    )

    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(9,6))
plt.plot(
    range(1, 11),
    wcss,
    marker='o'
)

plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.xticks(range(1, 11))
plt.show()

# Train K-Means with 5 Clusters
kmeans = KMeans(
    n_clusters=5,
    init='k-means++',
    random_state=42,
    n_init=10
)

df['Cluster'] = kmeans.fit_predict(X)

# Cluster Visualization
plt.figure(figsize=(10,7))

sns.scatterplot(
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    data=df,
    palette='Set1',
    s=100
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    s=300,
    marker='X',
    color='black',
    label='Centroids'
)

plt.title('Customer Segmentation using K-Means')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()

# Cluster Centers
print("\nCluster Centers:")
cluster_centers = pd.DataFrame(
    kmeans.cluster_centers_,
    columns=[
        'Annual Income (k$)',
        'Spending Score (1-100)'
    ]
)

display(cluster_centers)

# Number of Customers in Each Cluster
print("\nNumber of Customers in Each Cluster:")
print(df['Cluster'].value_counts().sort_index())

plt.figure(figsize=(8,5))
sns.countplot(x='Cluster', data=df)
plt.title('Number of Customers in Each Cluster')
plt.xlabel('Cluster')
plt.ylabel('Number of Customers')
plt.show()

# Cluster Analysis
print("\nCluster Analysis:")
cluster_analysis = df.groupby('Cluster')[
    [
        'Age',
        'Annual Income (k$)',
        'Spending Score (1-100)'
    ]
].mean().round(2)

cluster_analysis['Customer Count'] = (
    df['Cluster'].value_counts().sort_index()
)

display(cluster_analysis)

# Silhouette Score
silhouette = silhouette_score(
    X,
    df['Cluster']
)

print("\nSilhouette Score:", round(silhouette, 3))

# Final Dataset
print("\nFinal Dataset with Cluster Labels:")
display(df.head(20))
```
