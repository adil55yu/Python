Task no 1:

```python
sentence = input("Enter a sentence: ")

print("Uppercase:", sentence.upper())
print("Lowercase:", sentence.lower())
print("Title Case:", sentence.title())
print("Number of characters:", len(sentence))
print("Number of spaces:", sentence.count(" "))

print("Occurrences of each letter:")
for letter in "abcdefghijklmnopqrstuvwxyz":
    count = sentence.lower().count(letter)
    if count > 0:
        print(letter, ":", count)
```

Task no 4:

```python
numbers = [69, 12, 3, 20, 5, 4, 10, 67, 19, 99]

print("Original list:", numbers)

numbers.append(30)
print("After adding an element:", numbers)

numbers.insert(2, 100)
print("After inserting an element:", numbers)

numbers.remove(20)
print("After removing an element:", numbers)

numbers.sort()
print("After sorting:", numbers)

numbers.reverse()
print("After reversing:", numbers)

print("Final list:", numbers)
```
