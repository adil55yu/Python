```python
def encrypt(text, shift):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            new_char = chr((ord(ch) - base + shift) % 26 + base)
            result += new_char
        else:
            result += ch

    return result
```

```python
def decrypt(text, shift):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            new_char = chr((ord(ch) - base - shift) % 26 + base)
            result += new_char
        else:
            result += ch

    return result
```

```python
text = "THIS IS A SECRET"
shift = 2

cipher = encrypt(text, shift)
print("Encrypted:", cipher)

plain = decrypt(cipher, shift)
print("Decrypted:", plain)
```

```python
import string

def vigenere_skip_vowels_encrypt(plaintext, key):
    plaintext = plaintext.upper()
    key = key.upper()

    vowels = "AEIOU"
    result = []
    key_index = 0
    key_length = len(key)

    print("\nStep-by-step Mapping:\n")
    print("Char | Type      | Key Used | Operation | Result")

    for ch in plaintext:
        if ch in vowels:
            result.append(ch)
            print(f"{ch:4} | Vowel     |   -      | No change | {ch}")

        else:
            p = ord(ch) - ord('A')
            k = ord(key[key_index % key_length]) - ord('A')

            encrypted = chr((p + k) % 26 + ord('A'))
            result.append(encrypted)

            print(f"{ch:4} | Consonant |   {key[key_index % key_length]}      | "
                  f"({ch} + {key[key_index % key_length]}) | {encrypted}")

            key_index += 1

    ciphertext = "".join(result)
    return ciphertext


# Input
plaintext = "SECURITYLAB"
key = "KEY"

ciphertext = vigenere_skip_vowels_encrypt(plaintext, key)

print("\nFinal Ciphertext:", ciphertext)
```

```python
import math

def columnar_cipher_encrypt(plaintext, key):
    plaintext = plaintext.replace(" ", "").upper()
    key = key.upper()

    cols = len(key)
    rows = math.ceil(len(plaintext) / cols)


    matrix = [['' for _ in range(cols)] for _ in range(rows)]


    index = 0
    for i in range(rows):
        for j in range(cols):
            if index < len(plaintext):
                matrix[i][j] = plaintext[index]
                index += 1


    print("\nMatrix:")
    for row in matrix:
        print(row)


    key_order = sorted(list(enumerate(key)), key=lambda x: x[1])
    order = [i[0] for i in key_order]

    print("\nColumn Order (based on key):", order)


    ciphertext = ""
    for col in order:
        for row in range(rows):
            if matrix[row][col] != '':
                ciphertext += matrix[row][col]

    return ciphertext


plaintext = "NETWORKSECURITY"
key = "HACK"

result = columnar_cipher_encrypt(plaintext, key)
print("\nCiphertext:", result)
```

```python
import string

def caesar_decrypt(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            base = ord('A')
            result += chr((ord(ch) - base - shift) % 26 + base)
        else:
            result += ch
    return result


def score_english(text):

    common_words = ["THE", "AND", "TO", "OF", "IN", "IS", "IT", "YOU", "THAT"]
    score = 0
    words = text.split()
    for w in words:
        if w in common_words:
            score += 5
    return score


def find_caesar_key(ciphertext):
    ciphertext = ciphertext.upper()

    best_shift = 0
    best_score = -1
    best_plain = ""

    for shift in range(26):
        decrypted = caesar_decrypt(ciphertext, shift)
        score = score_english(decrypted)

        if score > best_score:
            best_score = score
            best_shift = shift
            best_plain = decrypted

    return best_shift, best_plain


ciphertext = "QEB NRFZH YOLTK CLU GRJMP LSBO QEB IXWV ALD"

shift, plaintext = find_caesar_key(ciphertext)

print("Detected Key (Shift):", shift)
print("Decrypted Text:")
print(plaintext)
```
