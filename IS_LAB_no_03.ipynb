Task 1 — Secure Text Processing System    
Design a program that:     
1. Takes a plaintext message input    
2. Converts message to uppercase    
3. Removes spaces   
4. Displays processed message    
5. Displays message length    
6. Displays characters one by one     
7. Displays ASCII value of each character    
8. Converts ASCII values back to characters    
9. Displays reconstructed message     

```python
message = input("Enter a plaintext message: ")

uppercase_message = message.upper()
processed_message = uppercase_message.replace(" ", "")

print("\nProcessed Message:", processed_message)
print("Message Length:", len(processed_message))

print("\nCharacters one by one:")
for char in processed_message:
    print(char)

ascii_values = []
print("\nASCII values:")
for char in processed_message:
    ascii_val = ord(char)
    ascii_values.append(ascii_val)
    print(f"{char} -> {ascii_val}")

reconstructed_message = ""
for value in ascii_values:
    reconstructed_message += chr(value)

print("\nReconstructed Message:", reconstructed_message)
```

Task 2 — Character Frequency Analyzer
Design a program that:
1. Takes message input
2. Removes spaces
3. Converts to uppercase
4. Counts occurrence of each character using dictionary
5. Displays frequency table
6. Displays most frequent character
7. Displays least frequent character

```python
message = input("Enter a message: ")

message = message.replace(" ", "").upper()

frequency = {}

for char in message:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("\nFrequency Table:")
for char, count in frequency.items():
    print(f"{char} -> {count}")

most_frequent = max(frequency, key=frequency.get)
least_frequent = min(frequency, key=frequency.get)

print("\nMost Frequent Character:", most_frequent, "->", frequency[most_frequent])
print("Least Frequent Character:", least_frequent, "->", frequency[least_frequent])
```

Task 3 — Alphabet Based Character Shifting System
Design a program that:
1. Takes plaintext input
2. Converts to uppercase
3. Removes spaces
4. Takes shift value input
5. Shifts characters using alphabet indexing
6. Uses modulo arithmetic
7. Displays shifted text

```python
plaintext = input("Enter plaintext: ")

text = plaintext.upper().replace(" ", "")

shift = int(input("Enter shift value: "))

shifted_text = ""

for char in text:
    if char.isalpha():
        index = ord(char) - ord('A')
        new_index = (index + shift) % 26
        new_char = chr(new_index + ord('A'))
        shifted_text += new_char
    else:
        shifted_text += char

print("\nShifted Text:", shifted_text)
```

Task 4 – Character Shift System    
Program should:   
1 Take plaintext input   
2 Convert to uppercase   
3 Remove spaces   
4 Take shift value   
5 Shift characters using modulo arithmetic   
6 Display result  

```python
 plaintext = input("Enter plaintext: ")

text = plaintext.upper().replace(" ", "")

shift = int(input("Enter shift value: "))

result = ""

for char in text:
    if char.isalpha():
        shifted = (ord(char) - ord('A') + shift) % 26
        result += chr(shifted + ord('A'))
    else:
        result += char

print("\nShifted Result:", result)
```

Task 5 — File Based Message Processor
Design a program that:
1. Reads plaintext from file
2. Converts to uppercase
3. Removes spaces
4. Displays processed text
5. Writes processed text into new file
6. Displays total characters processed

```python
input_file = "input.txt"
output_file = "output.txt"

try:
    with open(input_file, "r") as file:
        text = file.read()

    print("Original Text:")
    print(text)

    text_upper = text.upper()
    processed_text = text_upper.replace(" ", "")

    print("\nProcessed Text:")
    print(processed_text)

    with open(output_file, "w") as file:
        file.write(processed_text)

    total_characters = len(processed_text)
    print("\nTotal Characters Processed:", total_characters)

except FileNotFoundError:
    print("Error: Input file not found.")
```

Task 6 — Hash Generation System
Design a program that:
1. Takes message input
2. Converts message to bytes
3. Generates MD5 hash
4. Generates SHA-1 hash
5. Generates SHA-256 hash
6. Displays all hashes

```python
import hashlib

message = input("Enter message: ")

message_bytes = message.encode()

md5_hash = hashlib.md5(message_bytes).hexdigest()
sha1_hash = hashlib.sha1(message_bytes).hexdigest()
sha256_hash = hashlib.sha256(message_bytes).hexdigest()

print("\nMD5 Hash:", md5_hash)
print("SHA-1 Hash:", sha1_hash)
print("SHA-256 Hash:", sha256_hash)
```

Task 7 — Substitution Mapping System
Design a program that:
1. Creates dictionary mapping:
A → D
B → E
C → F
2. Takes plaintext input
3. Converts to uppercase
4. Replaces characters using dictionary
5. Displays substituted text

```python
mapping = {
    'A': 'D',
    'B': 'E',
    'C': 'F'
}

plaintext = input("Enter plaintext: ")

plaintext = plaintext.upper()

substituted_text = ""

for char in plaintext:
    if char in mapping:
        substituted_text += mapping[char]
    else:
        substituted_text += char

print("\nSubstituted Text:", substituted_text)
```
