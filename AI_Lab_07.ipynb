```python
import random

def create_board(n):
    board = [["*" for _ in range(n)] for _ in range(n)]
    queens_pos = []

    # Place one queen randomly in each column
    for col in range(n):
        row = random.randint(0, n - 1)
        board[row][col] = "Q"
        queens_pos.append((row, col))

    return board, queens_pos


def print_board(board):
    for row in board:
        print(" ".join(row))


def objective_function(queens_pos):
    conflicts = 0
    print("\nConflict Details:\n")

    for i in range(len(queens_pos)):
        for j in range(i + 1, len(queens_pos)):

            row_diff = abs(queens_pos[i][0] - queens_pos[j][0])
            col_diff = abs(queens_pos[i][1] - queens_pos[j][1])

            # Same row or same diagonal
            if row_diff == 0 or row_diff == col_diff:
                print(f"queens_pos[{i}][0] - queens_pos[{j}][0] = {row_diff}")
                print(f"queens_pos[{i}][1] - queens_pos[{j}][1] = {col_diff}")
                conflicts += 1

    return conflicts


def main():
    n = int(input("Enter the board size: "))

    # To match the screenshot
    print([[1]])

    board, queens_pos = create_board(n)

    print("\nQueen Positions:")
    print(queens_pos)

    print("\nBoard:\n")
    print_board(board)

    conflicts = objective_function(queens_pos)

    print("\nObjective Function", conflicts)


if __name__ == "__main__":
    main()
```
