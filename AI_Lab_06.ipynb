1. Develop a Model based reflex agent program in Python for the vacuum-cleaner-
world problem.

```python
rooms = ["A", "B", "C"]

location = input("Enter Initial Location (A/B/C): ").upper()

status = {}
for room in rooms:
    status[room] = int(input(f"Enter Status of Room {room} (0=Clean, 1=Dirty): "))

path_cost = 0

print("\nInitial Room Status:")
print(status)

while 1 in status.values():

    if status[location] == 1:
        print(f"Cleaning Room {location}")
        status[location] = 0
        path_cost += 1
    else:
        print(f"Room {location} is already Clean.")

    if 1 not in status.values():
        break

    current = rooms.index(location)

    if current < len(rooms) - 1:
        location = rooms[current + 1]
        print(f"Moving to Room {location}")
        path_cost += 1
    else:
        break

print("\nFinal Room Status:")
print(status)
print("Goal State Achieved!")
print("Total Path Cost =", path_cost)
```

3. Modify the vacuum cleaner program to count how many rooms were cleaned.

```python
# Vacuum Cleaner Agent with Room Counter

rooms = ["A", "B", "C"]

location = input("Enter Initial Location (A/B/C): ").upper()

status = {}
for room in rooms:
    status[room] = int(input(f"Enter Status of Room {room} (0=Clean, 1=Dirty): "))

path_cost = 0
cleaned_rooms = 0

while 1 in status.values():

    if status[location] == 1:
        print(f"Cleaning Room {location}")
        status[location] = 0
        cleaned_rooms += 1
        path_cost += 1
    else:
        print(f"Room {location} is already Clean.")

    if 1 not in status.values():
        break

    current = rooms.index(location)

    if current < len(rooms) - 1:
        location = rooms[current + 1]
        print(f"Moving to Room {location}")
        path_cost += 1
    else:
        break

print("\nFinal Room Status:")
print(status)
print("Total Rooms Cleaned =", cleaned_rooms)
print("Total Path Cost =", path_cost)
```
