Task 01 – RSA Key Generation   
Task Statement     
Design a Python program to generate RSA public and private keys.   
Requirements  
1. Input two prime numbers (p, q)
2. Compute n and φ(n)
3. Choose valid e
4. Compute d using modular inverse  
Output Must Include   
p = ...  
q = ...  
n = ...  
φ(n) = ...  
e = ...  
d = ...  
Public Key (e, n) = ...  
Private Key (d, n) = ...  

```python
def is_prime(num):
    if num < 2:
        return False
    i = 2
    while i * i <= num:
        if num % i == 0:
            return False
        i += 1
    return True

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def mod_inverse(e, phi):
    t, new_t = 0, 1
    r, new_r = phi, e

    while new_r != 0:
        quotient = r // new_r
        t, new_t = new_t, t - quotient * new_t
        r, new_r = new_r, r - quotient * new_r

    if r > 1:
        return None
    if t < 0:
        t += phi

    return t

p = int(input("Enter prime number p: "))
q = int(input("Enter prime number q: "))

if not (is_prime(p) and is_prime(q)):
    print("Error: Both numbers must be prime.")
    exit()

n = p * q
phi = (p - 1) * (q - 1)

e = 2
while e < phi:
    if gcd(e, phi) == 1:
        break
    e += 1

d = mod_inverse(e, phi)

print("\n--- RSA Key Generation Output ---")
print("p =", p)
print("q =", q)
print("n =", n)
print("φ(n) =", phi)
print("e =", e)
print("d =", d)
print("Public Key (e, n) =", (e, n))
print("Private Key (d, n) =", (d, n))
```

Task 02 – RSA Encryption  
Task Statement  
Encrypt a plaintext message using RSA public key.  
Requirements  
1. Input plaintext (numeric or ASCII)  
2. Use public key (e, n)  
3. Apply RSA encryption  
Output Must Include  
Plaintext (M) = ...  
Ciphertext (C) = ...

```python
plaintext = input("Enter plaintext: ")
e = int(input("Enter public key exponent (e): "))
n = int(input("Enter modulus (n): "))

if plaintext.isdigit():
    M = int(plaintext)
    C = pow(M, e, n)
    print("\nPlaintext (M) =", M)
    print("Ciphertext (C) =", C)
else:
    M = [ord(char) for char in plaintext]
    C = [pow(m, e, n) for m in M]
    print("\nPlaintext (M) =", M)
    print("Ciphertext (C) =", C)
```

Task 03 – RSA Decryption   
Task Statement   
Decrypt ciphertext using RSA private key.  
Requirements   
1. Input ciphertext      
2. Use private key (d, n)    
3. Recover original plaintext  
Output Must Include  
Ciphertext (C) = ...        
Decrypted Message (M) = ...  

```python
ciphertext = input("Enter ciphertext (space-separated numbers): ")
d = int(input("Enter private key exponent (d): "))
n = int(input("Enter modulus (n): "))

C = list(map(int, ciphertext.split()))
M_values = [pow(c, d, n) for c in C]
message = ''.join(chr(m) for m in M_values)

print("\nCiphertext (C) =", C)
print("Decrypted Message (M) =", message)
```

Task 04 – Full RSA Implementation  
Task Statement    
Develop a complete RSA system (key generation + encryption + decryption).  
Requirements  
1. Generate keys automatically  
2. Encrypt message  
3. Decrypt message  
4. Show all intermediate steps  
Output Must Include  
Keys:        
Public = ...  
Private = ...  
Encryption:  
M → C       
Decryption:   
C → M        
Verification:            
Original == Decrypted ✔   

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True


def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a


def mod_inverse(e, phi):
    t, new_t = 0, 1
    r, new_r = phi, e

    while new_r != 0:
        q = r // new_r
        t, new_t = new_t, t - q * new_t
        r, new_r = new_r, r - q * new_r

    if t < 0:
        t += phi
    return t


p = 17
q = 23

if not (is_prime(p) and is_prime(q)):
    print("Error: p and q must be prime")
    exit()

n = p * q
phi = (p - 1) * (q - 1)

e = 2
while e < phi:
    if gcd(e, phi) == 1:
        break
    e += 1

d = mod_inverse(e, phi)

print("Keys:")
print("Public Key =", (e, n))
print("Private Key =", (d, n))

message = input("\nEnter message: ")

M = []
C = []

for ch in message:
    M.append(ord(ch))
    C.append(pow(ord(ch), e, n))

print("\nEncryption:")
print("M →", M)
print("C →", C)

M_dec = []
for c in C:
    M_dec.append(pow(c, d, n))

decrypted = ""
for m in M_dec:
    decrypted += chr(m)

print("\nDecryption:")
print("C →", C)
print("M →", decrypted)

print("\nVerification:")
print("Original == Decrypted →", message == decrypted, "✔" if message == decrypted else "✘")
```

Task 05 – RSA Factorization Attack  
Task Statement   
Perform a factorization attack on a weak RSA implementation to recover the original plaintext.  
Given Information:   
text                   
Intercepted Public Key:  
e = 11  
n = 91                  
Intercepted Ciphertext:                                  
[32, 2, 34, 44, 44, 2, 27, 50, 51, 52, 2, 44, 50, 51, 52]   
Notes:                                               
- Encryption scheme: A=1, B=2, ..., Z=26, Space=0    
- Each number represents one character encrypted separately  
- Each character M was encrypted using: C = M^e mod n   
Requirements:                                    
1. Factor n to find p and q (show your work)   
2. Calculate φ(n)            
3. Calculate private key d  
4. Decrypt the ciphertext      
5. Reveal the hidden message        
Output Must Include:         
text                    
Step 1 - Factorization:        
n = 91                             
Factors found: p = ... , q = ...     
Step 2 - Calculate φ(n):     
φ(n) = (p-1)(q-1) = ...               
Step 3 - Calculate Private Key:         
e = 11                               
d = ... (where d × 11 ≡ 1 mod φ(n))       
Step 4 - Decryption:                                                   
Ciphertext: [32, 2, 34, 44, 44, 2, 27, 50, 51, 52, 2, 44, 50, 51, 52]
Decrypted values: [..., ..., ...]         
Recovered plaintext: "..."         
Verification:                                                               
Re-encrypting plaintext with public key should match original ciphertext ✓
Hint: To factor 91, try dividing by primes: 2, 3, 5, 7...    

```python
e = 11
n = 91
ciphertext = [32, 2, 34, 44, 44, 2, 27, 50, 51, 52, 2, 44, 50, 51, 52]

p, q = 7, 13
phi = (p - 1) * (q - 1)

def mod_inverse(e, phi):
    for d in range(1, phi):
        if (d * e) % phi == 1:
            return d

d = mod_inverse(e, phi)

decrypted = [(c ** d) % n for c in ciphertext]

plaintext = ""
for num in decrypted:
    if num == 0:
        plaintext += " "
    else:
        plaintext += chr(num + 64)

print("p =", p, "q =", q)
print("phi =", phi)
print("d =", d)
print("Decrypted values:", decrypted)
print("Plaintext:", plaintext)
```
