Task 1 – Columnar Cipher Encryption Program
Design a Python program that:
1. Accepts plaintext input
2. Accepts a keyword
3. Removes spaces from plaintext
4. Converts plaintext to uppercase
5. Determines column order based on sorted key
6. Arranges plaintext into a matrix
7. Adds padding characters if necessary
8. Generates ciphertext by reading columns in sorted key order

```python
plaintext = input("Enter Plaintext: ")
key = input("Enter Keyword: ")

plaintext = plaintext.replace(" ", "").upper()
key = key.upper()

key_order = sorted(list(enumerate(key)), key=lambda x: x[1])
order = [i[0] for i in key_order]

columns = len(key)
rows = len(plaintext) // columns
if len(plaintext) % columns != 0:
    rows += 1

padding = rows * columns - len(plaintext)
plaintext += 'X' * padding

matrix = []
index = 0
for r in range(rows):
    row = []
    for c in range(columns):
        row.append(plaintext[index])
        index += 1
    matrix.append(row)

print("\nMatrix:")
for row in matrix:
    print(" ".join(row))

ciphertext = ""
for col in order:
    for r in range(rows):
        ciphertext += matrix[r][col]

print("\nCiphertext:", ciphertext)
```

Task 2 – Columnar Cipher Decryption Program
Design a Python program that:
1. Accepts ciphertext
2. Accepts the encryption key
3. Calculates number of rows and columns
4. Reconstructs the ciphertext matrix
5. Rearranges columns back to original key order
6. Reads plaintext row-wise
7. Removes padding characters if present

```python
ciphertext = input("Enter Ciphertext: ")
key = input("Enter Keyword: ")

ciphertext = ciphertext.replace(" ", "").upper()
key = key.upper()

columns = len(key)
rows = len(ciphertext) // columns

key_order = sorted(list(enumerate(key)), key=lambda x: x[1])
order = [i[0] for i in key_order]

matrix = [['' for _ in range(columns)] for _ in range(rows)]

index = 0
for col in order:
    for r in range(rows):
        matrix[r][col] = ciphertext[index]
        index += 1

print("\nMatrix:")
for row in matrix:
    print(" ".join(row))

plaintext = ""
for r in range(rows):
    for c in range(columns):
        plaintext += matrix[r][c]

plaintext = plaintext.rstrip('X')

print("\nPlaintext:", plaintext)
```
